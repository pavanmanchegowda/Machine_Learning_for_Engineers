
# $\mathrm{1.0\ Introduction\ to\ Matplotlib}$

## $\mathrm{1.1\ What\ is\ Matplotlib?}$

$\displaystyle
\begin{aligned}
&\text{Matplotlib is an open-source Python library for data visualization.}\\[5pt]
&\text{It is used to create graphs and plots from numerical data.}\\[5pt]

$

\begin{aligned}


# $\mathrm{2.0\ Creating\ a\ Line\ Plot}$

## $\mathrm{2.1\ What\ is\ a\ Line\ Plot?}$

$\displaystyle
\begin{aligned}
&\text{A line plot represents data points connected by straight lines.}\\[5pt]
&\text{It is used to visualize how one variable changes with another.}\\[5pt]
&\text{For example, temperature variation with time.}
\end{aligned}
$

## $\mathrm{2.2\ Important\ Plotting\ Functions}$

$\displaystyle
\begin{aligned}
&\texttt{plt.plot(): Create a line plot.}\\[5pt]
&\texttt{plt.title(): Set the plot title.}\\[5pt]
&\texttt{plt.xlabel(): Label the x-axis.}\\[5pt]
&\texttt{plt.ylabel(): Label the y-axis.}\\[5pt]
&\texttt{plt.show(): Display the plot.}
\end{aligned}
$

## $\mathrm{2.3\ Line\ Plot\ Features}$

$\displaystyle
\begin{aligned}
&\texttt{color: Set the line color.}\\[5pt]
&\texttt{marker: Set the symbol used for data points.}\\[5pt]
&\texttt{markersize: Set the size of the markers.}\\[5pt]
&\texttt{linestyle: Set the line pattern.}\\[5pt]
&\texttt{linewidth: Set the thickness of the line.}\\[5pt]
&\texttt{label: Assign a name to the plotted data.}
\end{aligned}
$

## $\mathrm{2.4\ Common\ Marker\ Styles}$

$\displaystyle
\begin{aligned}
&\texttt{o: Circle}\\[5pt]
&\texttt{s: Square}\\[5pt]
&\texttt{*: Star}\\[5pt]
&\texttt{\^{}: Triangle}\\[5pt]
&\texttt{x: Cross}\\[5pt]
&\texttt{+: Plus}
\end{aligned}
$

## $\mathrm{2.5\ Common\ Line\ Styles}$

$\displaystyle
\begin{aligned}
&\texttt{-: Solid line}\\[5pt]
&\texttt{--: Dashed line}\\[5pt]
&\texttt{:: Dotted line}\\[5pt]
&\texttt{-.: Dash-dot line}
\end{aligned}
$


```python

import numpy as np


# 1. Create sample data
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([1, 3, 5, 7, 9, 11])

# 2. Create a line plot
plt.plot(x, y, color="black", marker="*")

# 3. Add title and axis labels
plt.title("My First Plot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

# 4. Display the plot
plt.show()

```


# $\mathrm{3.0\ Customizing\ Plots\ Using\ rcParams}$

## $\mathrm{3.1\ What\ is\ rcParams?}$

$\displaystyle
\begin{aligned}
&\text{rcParams is a configuration dictionary provided by Matplotlib.}\\[5pt]
&\text{It controls the default appearance of plots.}\\[5pt]
&\text{Settings can be applied globally to all subsequent figures.}
\end{aligned}
$

## $\mathrm{3.2\ Important\ rcParams\ Settings}$

$\displaystyle
\begin{aligned}
&\texttt{font.family: Set the default font family.}\\[5pt]
&\texttt{font.size: Set the default font size.}\\[5pt]
&\texttt{axes.titlesize: Set the title font size.}\\[5pt]
&\texttt{axes.labelsize: Set the axis label font size.}\\[5pt]
&\texttt{xtick.labelsize: Set the x-axis tick label size.}\\[5pt]
&\texttt{ytick.labelsize: Set the y-axis tick label size.}\\[5pt]
&\texttt{text.usetex: Enable external LaTeX rendering.}\\[5pt]
&\texttt{mathtext.fontset: Set the mathematical font style.}
\end{aligned}
$

## $\mathrm{3.3\ LaTeX\ Formatting}$

$\displaystyle
\begin{aligned}
&\text{Matplotlib supports mathematical expressions in plot titles and labels.}\\[5pt]
&\text{Mathematical expressions are enclosed between dollar signs.}\\[5pt]
&\text{For example, } y=x^2 \text{ can be displayed using LaTeX-style notation.}\\[5pt]
&\text{External LaTeX rendering requires a LaTeX installation.}
\end{aligned}
$


```python


from matplotlib import rcParams

rcParams["text.usetex"] = False
rcParams["mathtext.fontset"] = "cm"
rcParams["font.family"] = "serif"
rcParams["font.serif"] = ["STIXGeneral"]
rcParams["font.size"] = 12
rcParams["axes.titlesize"] = 16
rcParams["axes.labelsize"] = 14

```

```python

# Sample data
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([1, 3, 5, 7, 9, 11])

plt.plot(x, y, color="darkred", marker="o", markersize=6)

# LaTeX-style title and axis labels
plt.title(r"$\mathrm{My\ First\ Plot}$")
plt.xlabel(r"$\mathrm{X-Axis}$")
plt.ylabel(r"$\mathrm{Y-Axis}$")

plt.grid(True, linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# 1. Create sample data (5 parallel lines)
x = np.linspace(0, 10, 21)

data = [
    2*x + 1,
    2*x + 4,
    2*x + 7,
    2*x + 10,
    2*x + 13
]

colors = ["red", "green", "blue", "orange", "purple"]
markers = [".", "*", "o", "s", "^"]
styles = ["-", "--", ":", "-.", "-"]

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Regular plot (intentionally poorly formatted)
for i in range(5):
    ax[0].plot(x, data[i],
               color=colors[i],
               linestyle=styles[i],
               marker=markers[i])

ax[0].set_title("Comparison of Lines")
ax[0].set_xlabel("X-axis")
ax[0].set_ylabel("Y-axis")

# RIGHT: Improved LaTeX-style plot
for i in range(5):
    ax[1].plot(x, data[i],
               color=colors[i],
               linestyle=styles[i],
               marker=markers[i],
               markersize=6,
               markevery=2,
               linewidth=1.8,
               label=rf"$\mathrm{{Line\ {i+1}}}$")

ax[1].set_title(r"$\mathrm{Comparison\ of\ Lines}$",
                fontsize=16, pad=12)

ax[1].set_xlabel(r"$\mathrm{X-axis}$",
                 fontsize=14, labelpad=8)

ax[1].set_ylabel(r"$\mathrm{Y-axis}$",
                 fontsize=14, labelpad=8)

# Grid
ax[1].grid(True, linestyle="--", alpha=0.25)

# Legend
ax[1].legend(
    loc="upper left",
    frameon=True,
    fancybox=True,
    framealpha=0.95,
    edgecolor="black",
    fontsize=10
)

# Tick formatting
ax[1].tick_params(axis="both",
                  direction="in",
                  length=5,
                  width=1)

# Complete black box around the LaTeX plot
for spine in ax[1].spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1.2)

# 3. Display
plt.tight_layout(w_pad=2)
plt.show()
```


# $\mathrm{5.0\ 3D\ Surface\ Plots\ and\ Colormaps}$

## $\mathrm{5.1\ What\ is\ a\ Surface\ Plot?}$

$\displaystyle
\begin{aligned}
&\text{A 3D surface plot represents a function of two independent variables.}\\[5pt]
&\text{The X-axis and Y-axis define the surface coordinates.}\\[5pt]
&\text{The Z-axis represents the function value.}\\[5pt]
&\text{A colormap assigns colors to numerical values.}
\end{aligned}
$

The mathematical representation is:

$$
z=f(x,y)
$$

## $\mathrm{5.2\ Creating\ a\ Meshgrid}$

$\displaystyle
\begin{aligned}
&\text{A meshgrid converts one-dimensional coordinate arrays into}\\
&\text{two-dimensional coordinate matrices.}\\[5pt]
&\text{The NumPy function meshgrid() is used for this purpose.}
\end{aligned}
$

$$
X,Y=\operatorname{meshgrid}(x,y)
$$

## $\mathrm{5.3\ What\ is\ a\ Colormap?}$

$\displaystyle
\begin{aligned}
&\text{A colormap maps numerical values to different colors.}\\[5pt]
&\text{It helps visualize variations in surface height or magnitude.}\\[5pt]
&\text{Matplotlib provides several predefined colormaps.}
\end{aligned}
$

## $\mathrm{5.4\ Common\ Matplotlib\ Colormaps}$

$\displaystyle
\begin{aligned}
&\text{viridis, plasma, inferno, magma}\\[5pt]
&\text{cividis, coolwarm, jet, turbo}\\[5pt]
&\text{Spectral, RdBu, terrain, cool}
\end{aligned}
$

## $\mathrm{5.5\ Example\ Surface\ Function}$

For the following example, we consider:

$$
z=\sin\left(\sqrt{x^2+y^2}\right)
$$

$\displaystyle
\begin{aligned}
&\text{The same surface is plotted using 12 different colormaps.}\\[5pt]
&\text{A 4-by-3 subplot arrangement is used for comparison.}\\[5pt]
&\text{Only the colormap changes; the numerical data remains identical.}
\end{aligned}
$


```python

from matplotlib import colormaps

# 1. Create 3D data
x = np.linspace(-4, 4, 100)
y = np.linspace(-4, 4, 100)

X, Y = np.meshgrid(x, y)

Z = np.sin(np.sqrt(X**2 + Y**2))

# 2. Select 12 colormaps
cmaps = [
    "viridis", "plasma", "inferno",
    "magma", "cividis", "coolwarm",
    "jet", "turbo", "Spectral",
    "RdBu", "terrain", "cool"
]

# 3. Create a 4x3 figure
fig, ax = plt.subplots(
    4, 3,
    figsize=(13, 15),
    subplot_kw={"projection": "3d"}
)

# 4. Plot the same surface with different colormaps
for axis, cmap in zip(ax.flat, cmaps):

    axis.plot_surface(
        X, Y, Z,
        cmap=colormaps[cmap],
        edgecolor="none"
    )

    axis.set_title(cmap, fontsize=12)
    axis.set_xlabel("X-axis")
    axis.set_ylabel("Y-axis")
    axis.set_zlabel("Z-axis")

# 5. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.1\ Bar\ Plot}$

$\displaystyle
\begin{aligned}
&\text{A bar plot represents numerical values using rectangular bars.}\\[5pt]
&\text{It is used to compare values across different categories.}\\[5pt]
&\text{The height of each bar represents its numerical value.}\\[5pt]
&\text{Matplotlib function: plt.bar().}
\end{aligned}
$

## $\mathrm{Example:\ Product\ Sales}$

$\displaystyle
\begin{aligned}
&\text{Consider five products with different monthly sales.}\\[5pt]
&\text{The X-axis represents the product categories.}\\[5pt]
&\text{The Y-axis represents the number of units sold.}
\end{aligned}
$


```python

# 1. Create sample data
products = ["A", "B", "C", "D", "E"]
sales = [120, 180, 95, 220, 150]

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic plot
ax[0].bar(products, sales)
ax[0].set_title("Product Sales")
ax[0].set_xlabel("Products")
ax[0].set_ylabel("Units Sold")

# RIGHT: Improved LaTeX-style plot
bars = ax[1].bar(
    products, sales,
    color="royalblue",
    edgecolor="black",
    width=0.6
)

ax[1].set_title(r"$\mathrm{Product\ Sales}$", fontsize=16)
ax[1].set_xlabel(r"$\mathrm{Products}$", fontsize=13)
ax[1].set_ylabel(r"$\mathrm{Units\ Sold}$", fontsize=13)

ax[1].grid(axis="y", linestyle="--", alpha=0.3)
ax[1].set_axisbelow(True)
ax[1].bar_label(bars, padding=3)

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.2\ Histogram}$

$\displaystyle
\begin{aligned}
&\text{A histogram represents the frequency distribution of numerical data.}\\[5pt]
&\text{The data is divided into intervals called bins.}\\[5pt]
&\text{The height of each bar represents the number of observations in a bin.}\\[5pt]
&\text{Unlike bar plots, histograms represent continuous numerical data.}\\[5pt]
&\text{Matplotlib function: plt.hist().}
\end{aligned}
$

## $\mathrm{Example:\ Customer\ Spending}$

$\displaystyle
\begin{aligned}
&\text{Consider the purchase amounts of 500 customers.}\\[5pt]
&\text{The X-axis represents the amount spent by customers.}\\[5pt]
&\text{The Y-axis represents the number of customers.}\\[5pt]
&\text{The histogram helps identify common spending ranges.}
\end{aligned}
$


```python

# 1. Create sample data
rng = np.random.default_rng(42)
spending = rng.normal(loc=100, scale=25, size=500)

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic histogram
ax[0].hist(spending)
ax[0].set_title("Customer Spending")
ax[0].set_xlabel("Amount Spent")
ax[0].set_ylabel("Frequency")

# RIGHT: Improved LaTeX-style histogram
ax[1].hist(
    spending,
    bins=20,
    color="royalblue",
    edgecolor="black",
    alpha=0.85
)

ax[1].set_title(r"$\mathrm{Customer\ Spending}$", fontsize=16)
ax[1].set_xlabel(r"$\mathrm{Amount\ Spent}$", fontsize=13)
ax[1].set_ylabel(r"$\mathrm{Frequency}$", fontsize=13)

ax[1].grid(axis="y", linestyle="--", alpha=0.3)
ax[1].set_axisbelow(True)

# Complete black box
for spine in ax[1].spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1.2)

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.3\ Scatter\ Plot}$

$\displaystyle
\begin{aligned}
&\text{A scatter plot represents individual data points on a 2D plane.}\\[5pt]
&\text{It is used to identify relationships between two numerical variables.}\\[5pt]
&\text{Unlike line plots, the data points are not connected by lines.}\\[5pt]
&\text{Matplotlib function: plt.scatter().}
\end{aligned}
$

## $\mathrm{Example:\ Advertising\ and\ Sales}$

$\displaystyle
\begin{aligned}
&\text{Consider the advertising expenditure and sales revenue of a company.}\\[5pt]
&\text{The X-axis represents advertising expenditure.}\\[5pt]
&\text{The Y-axis represents sales revenue.}\\[5pt]
&\text{The scatter plot helps identify the relationship between them.}
\end{aligned}
$


```python

# 1. Create sample data
rng = np.random.default_rng(42)

advertising = np.linspace(10, 100, 50)
sales = 2.5 * advertising + 40 + rng.normal(0, 20, 50)

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic scatter plot
ax[0].scatter(advertising, sales)
ax[0].set_title("Advertising vs Sales")
ax[0].set_xlabel("Advertising Expenditure")
ax[0].set_ylabel("Sales Revenue")

# RIGHT: Improved LaTeX-style scatter plot
ax[1].scatter(
    advertising, sales,
    color="royalblue",
    marker="o",
    s=45,
    edgecolor="black",
    linewidth=0.5,
    alpha=0.85,
    label=r"$\mathrm{Sales\ Data}$"
)

ax[1].set_title(r"$\mathrm{Advertising\ vs\ Sales}$",
                fontsize=16, pad=12)

ax[1].set_xlabel(r"$\mathrm{Advertising\ Expenditure}$",
                 fontsize=13)

ax[1].set_ylabel(r"$\mathrm{Sales\ Revenue}$",
                 fontsize=13)

ax[1].grid(True, linestyle="--", alpha=0.3)
ax[1].set_axisbelow(True)
ax[1].legend(frameon=True, edgecolor="black")

# Complete black box
for spine in ax[1].spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1.2)

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.4\ Contour\ Plot}$

$\displaystyle
\begin{aligned}
&\text{A contour plot represents a three-dimensional function on a 2D plane.}\\[5pt]
&\text{Contour lines connect points having the same numerical value.}\\[5pt]
&\text{It is useful for visualizing spatial variations of scalar quantities.}\\[5pt]
&\text{Matplotlib function: plt.contour().}
\end{aligned}
$

## $\mathrm{Example:\ Two\ Dimensional\ Function}$

Consider the function:

$$
z=\sin(x)\cos(y)
$$

$\displaystyle
\begin{aligned}
&\text{The X-axis and Y-axis represent spatial coordinates.}\\[5pt]
&\text{Each contour line represents a constant value of z.}\\[5pt]
&\text{Different contour levels help identify variations in the function.}
\end{aligned}
$


```python

# 1. Create sample data
x = np.linspace(-4, 4, 100)
y = np.linspace(-4, 4, 100)

X, Y = np.meshgrid(x, y)
Z = np.sin(X) * np.cos(Y)

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic contour plot
ax[0].contour(X, Y, Z)
ax[0].set_title("Contour Plot")
ax[0].set_xlabel("X-axis")
ax[0].set_ylabel("Y-axis")

# RIGHT: Improved LaTeX-style contour plot
contours = ax[1].contour(
    X, Y, Z,
    levels=15,
    cmap="viridis",
    linewidths=1.2
)

# Label contour values
ax[1].clabel(contours, inline=True, fontsize=8, fmt="%.2f")

ax[1].set_title(r"$\mathrm{Contour\ Plot}$",
                fontsize=16, pad=12)

ax[1].set_xlabel(r"$\mathrm{X-axis}$", fontsize=13)
ax[1].set_ylabel(r"$\mathrm{Y-axis}$", fontsize=13)

ax[1].grid(True, linestyle="--", alpha=0.2)

# Colorbar
cbar = fig.colorbar(contours, ax=ax[1], shrink=0.85, pad=0.04)
cbar.set_label(r"$z=\sin(x)\cos(y)$", fontsize=12)

# Complete black box
for spine in ax[1].spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1.2)

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.5\ Heatmap}$

$\displaystyle
\begin{aligned}
&\text{A heatmap represents numerical values using different colors.}\\[5pt]
&\text{It is useful for visualizing matrices and two-dimensional datasets.}\\[5pt]
&\text{Each color corresponds to a particular numerical value.}\\[5pt]
&\text{A colorbar indicates the relationship between colors and values.}\\[5pt]
&\text{Matplotlib function: plt.imshow().}
\end{aligned}
$

## $\mathrm{Example:\ Monthly\ Product\ Sales}$

$\displaystyle
\begin{aligned}
&\text{Consider the monthly sales of five different products.}\\[5pt]
&\text{The X-axis represents the months.}\\[5pt]
&\text{The Y-axis represents the products.}\\[5pt]
&\text{The color intensity represents the number of units sold.}
\end{aligned}
$


```python

# 1. Create sample data
sales = np.array([
    [120, 150, 180, 160, 200, 220],
    [90, 110, 130, 145, 170, 190],
    [200, 210, 190, 230, 250, 270],
    [80, 100, 120, 110, 140, 160],
    [150, 170, 160, 190, 210, 240]
])

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
products = ["A", "B", "C", "D", "E"]

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic heatmap
ax[0].imshow(sales)
ax[0].set_title("Monthly Product Sales")
ax[0].set_xlabel("Months")
ax[0].set_ylabel("Products")

# RIGHT: Improved LaTeX-style heatmap
heatmap = ax[1].imshow(
    sales,
    cmap="YlGnBu",
    aspect="auto",
    interpolation="nearest"
)

ax[1].set_title(r"$\mathrm{Monthly\ Product\ Sales}$",
                fontsize=16, pad=12)

ax[1].set_xlabel(r"$\mathrm{Months}$", fontsize=13)
ax[1].set_ylabel(r"$\mathrm{Products}$", fontsize=13)

# Category labels
ax[1].set_xticks(np.arange(len(months)), labels=months)
ax[1].set_yticks(np.arange(len(products)), labels=products)

# Display numerical values inside each cell
for i in range(len(products)):
    for j in range(len(months)):
        ax[1].text(
            j, i, str(sales[i, j]),
            ha="center", va="center",
            fontsize=9,
            color="white" if sales[i, j] > 180 else "black"
        )

# Colorbar
cbar = fig.colorbar(heatmap, ax=ax[1], shrink=0.85, pad=0.04)
cbar.set_label(r"$\mathrm{Units\ Sold}$", fontsize=12)

# Complete black box
for spine in ax[1].spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1.2)

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.6\ Quiver\ Plot}$

$\displaystyle
\begin{aligned}
&\text{A quiver plot represents a vector field using arrows.}\\[5pt]
&\text{The direction of each arrow indicates the vector direction.}\\[5pt]
&\text{The arrow length can represent the vector magnitude.}\\[5pt]
&\text{It is useful for visualizing flow, motion, and directional data.}\\[5pt]
&\text{Matplotlib function: plt.quiver().}
\end{aligned}
$

## $\mathrm{Example:\ Two\ Dimensional\ Vector\ Field}$

Consider the vector field:

$$
\vec{V}(x,y)=
\begin{bmatrix}
-y\\
x
\end{bmatrix}
$$

The vector components are:

$$
u(x,y)=-y,\qquad v(x,y)=x
$$

The vector magnitude is:

$$
|\vec{V}|=\sqrt{u^2+v^2}
$$

$\displaystyle
\begin{aligned}
&\text{The X-axis and Y-axis represent spatial coordinates.}\\[5pt]
&\text{The arrows indicate a rotational vector field.}\\[5pt]
&\text{The improved plot uses colors to represent vector magnitude.}
\end{aligned}
$


```python

# 1. Create sample data
x = np.linspace(-3, 3, 15)
y = np.linspace(-3, 3, 15)

X, Y = np.meshgrid(x, y)

# Vector components
U = -Y
V = X

# Vector magnitude
M = np.sqrt(U**2 + V**2)

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic quiver plot
ax[0].quiver(X, Y, U, V)
ax[0].set_title("Quiver Plot")
ax[0].set_xlabel("X-axis")
ax[0].set_ylabel("Y-axis")

# RIGHT: Improved LaTeX-style quiver plot
quiver = ax[1].quiver(
    X, Y, U, V, M,
    cmap="viridis",
    angles="xy",
    scale_units="xy",
    scale=10,
    width=0.004,
    pivot="mid"
)

ax[1].set_title(r"$\mathrm{Quiver\ Plot}$",
                fontsize=16, pad=12)

ax[1].set_xlabel(r"$\mathrm{X-axis}$", fontsize=13)
ax[1].set_ylabel(r"$\mathrm{Y-axis}$", fontsize=13)

ax[1].set_aspect("equal")
ax[1].grid(True, linestyle="--", alpha=0.25)

# Colorbar
cbar = fig.colorbar(quiver, ax=ax[1], shrink=0.85, pad=0.04)
cbar.set_label(r"$|\vec{V}|$", fontsize=12)

# Complete black box
for spine in ax[1].spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1.2)

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.7\ Streamline\ Plot}$

$\displaystyle
\begin{aligned}
&\text{A streamline plot visualizes the direction of a vector field.}\\[5pt]
&\text{Streamlines are curves tangent to the local velocity vectors.}\\[5pt]
&\text{They help visualize flow patterns and rotational motion.}\\[5pt]
&\text{Colors can represent the magnitude of the velocity field.}\\[5pt]
&\text{Matplotlib function: plt.streamplot().}
\end{aligned}
$

## $\mathrm{Example:\ Two\ Dimensional\ Rotational\ Flow}$

Consider the velocity field:

$$
\vec{V}(x,y)=
\begin{bmatrix}
-y\\
x
\end{bmatrix}
$$

The velocity components are:

$$
u(x,y)=-y,\qquad v(x,y)=x
$$

The velocity magnitude is:

$$
|\vec{V}|=\sqrt{u^2+v^2}
$$

$\displaystyle
\begin{aligned}
&\text{The X-axis and Y-axis represent spatial coordinates.}\\[5pt]
&\text{The streamlines form circular paths around the origin.}\\[5pt]
&\text{The improved plot uses a colormap to represent velocity magnitude.}
\end{aligned}
$


```python

# 1. Create sample data
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

X, Y = np.meshgrid(x, y)

# Velocity components
U = -Y
V = X

# Velocity magnitude
M = np.sqrt(U**2 + V**2)

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic streamline plot
ax[0].streamplot(X, Y, U, V)
ax[0].set_title("Streamline Plot")
ax[0].set_xlabel("X-axis")
ax[0].set_ylabel("Y-axis")

# RIGHT: Improved LaTeX-style streamline plot
stream = ax[1].streamplot(
    X, Y, U, V,
    color=M,
    cmap="viridis",
    density=1.5,
    linewidth=1.5,
    arrowsize=1.2
)

ax[1].set_title(r"$\mathrm{Streamline\ Plot}$",
                fontsize=16, pad=12)

ax[1].set_xlabel(r"$\mathrm{X-axis}$", fontsize=13)
ax[1].set_ylabel(r"$\mathrm{Y-axis}$", fontsize=13)

ax[1].set_aspect("equal")
ax[1].grid(True, linestyle="--", alpha=0.2)

# Colorbar
cbar = fig.colorbar(stream.lines, ax=ax[1], shrink=0.85, pad=0.04)
cbar.set_label(r"$|\vec{V}|$", fontsize=12)

# Complete black box
for spine in ax[1].spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1.2)

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.8\ Wireframe\ Plot}$

$\displaystyle
\begin{aligned}
&\text{A wireframe plot represents a 3D surface using intersecting lines.}\\[5pt]
&\text{It helps visualize the shape and structure of a surface.}\\[5pt]
&\text{Unlike surface plots, the regions between mesh lines are not filled.}\\[5pt]
&\text{The mesh density controls the number of displayed lines.}\\[5pt]
&\text{Matplotlib function: plot\_wireframe().}
\end{aligned}
$

## $\mathrm{Example:\ Three\ Dimensional\ Surface}$

Consider the function:

$$
z=\sin\left(\sqrt{x^2+y^2}\right)
$$

$\displaystyle
\begin{aligned}
&\text{The X-axis and Y-axis define the surface coordinates.}\\[5pt]
&\text{The Z-axis represents the function value.}\\[5pt]
&\text{The wireframe displays the surface using a structured mesh.}
\end{aligned}
$


```python

# 1. Create sample data
x = np.linspace(-4, 4, 60)
y = np.linspace(-4, 4, 60)

X, Y = np.meshgrid(x, y)
Z = np.sin(np.sqrt(X**2 + Y**2))

# 2. Create 1x2 subplots
fig, ax = plt.subplots(
    1, 2,
    figsize=(11, 4.5),
    subplot_kw={"projection": "3d"}
)

# LEFT: Basic wireframe plot
ax[0].plot_wireframe(X, Y, Z)

ax[0].set_title("Wireframe Plot")
ax[0].set_xlabel("X-axis")
ax[0].set_ylabel("Y-axis")
ax[0].set_zlabel("Z-axis")

# RIGHT: Improved LaTeX-style wireframe plot
ax[1].plot_wireframe(
    X, Y, Z,
    rstride=3,
    cstride=3,
    color="royalblue",
    linewidth=0.8
)

ax[1].set_title(r"$\mathrm{Wireframe\ Plot}$",
                fontsize=16, pad=12)

ax[1].set_xlabel(r"$\mathrm{X-axis}$", fontsize=12, labelpad=8)
ax[1].set_ylabel(r"$\mathrm{Y-axis}$", fontsize=12, labelpad=8)
ax[1].set_zlabel(r"$\mathrm{Z-axis}$", fontsize=12, labelpad=8)

ax[1].view_init(elev=30, azim=45)
ax[1].set_box_aspect((1, 1, 0.6))

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.9\ Pie\ Chart}$

$\displaystyle
\begin{aligned}
&\text{A pie chart represents the proportions of different categories.}\\[5pt]
&\text{Each sector represents a percentage of the total.}\\[5pt]
&\text{The sum of all sectors is equal to 100 percent.}\\[5pt]
&\text{It is useful for comparing relative contributions.}\\[5pt]
&\text{Matplotlib function: plt.pie().}
\end{aligned}
$

## $\mathrm{Example:\ Company\ Market\ Share}$

$\displaystyle
\begin{aligned}
&\text{Consider the market shares of five companies.}\\[5pt]
&\text{Each sector represents the market share of one company.}\\[5pt]
&\text{The improved chart displays percentages and category labels.}
\end{aligned}
$


```python

# 1. Create sample data
companies = ["A", "B", "C", "D", "E"]
market_share = [35, 25, 20, 12, 8]

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic pie chart
ax[0].pie(market_share)
ax[0].set_title("Company Market Share")

# RIGHT: Improved LaTeX-style pie chart
colors = ["royalblue", "cornflowerblue", "mediumseagreen",
          "gold", "lightcoral"]

wedges, texts, autotexts = ax[1].pie(
    market_share,
    labels=companies,
    colors=colors,
    autopct="%1.1f%%",
    startangle=90,
    explode=[0.06, 0, 0, 0, 0],
    shadow=True,
    wedgeprops={"edgecolor": "black", "linewidth": 0.8},
    textprops={"fontsize": 11},
    pctdistance=0.7
)

ax[1].set_title(r"$\mathrm{Company\ Market\ Share}$",
                fontsize=16, pad=12)

ax[1].legend(
    wedges,
    companies,
    title=r"$\mathrm{Companies}$",
    loc="center left",
    bbox_to_anchor=(1, 0.5),
    frameon=True,
    edgecolor="black"
)

ax[1].set_aspect("equal")

# 3. Display
plt.tight_layout()
plt.show()
```


# $\mathrm{6.10\ Box\ Plot}$

$\displaystyle
\begin{aligned}
&\text{A box plot summarizes the distribution of numerical data.}\\[5pt]
&\text{The box extends from the first quartile (Q1) to the third quartile (Q3).}\\[5pt]
&\text{The line inside the box represents the median (Q2).}\\[5pt]
&\text{Whiskers extend to the most extreme values within 1.5 times the IQR.}\\[5pt]
&\text{Points beyond the whiskers are displayed as potential outliers.}\\[5pt]
&\text{Matplotlib function: plt.boxplot().}
\end{aligned}
$

## $\mathrm{Example:\ Customer\ Spending\ Across\ Stores}$

$\displaystyle
\begin{aligned}
&\text{Consider customer spending at four different stores.}\\[5pt]
&\text{The X-axis represents the stores.}\\[5pt]
&\text{The Y-axis represents customer spending.}\\[5pt]
&\text{The box plots compare medians, variability, and potential outliers.}
\end{aligned}
$

The interquartile range is:

$$
\mathrm{IQR}=Q_3-Q_1
$$


```python

# 1. Create sample data
rng = np.random.default_rng(42)

store_a = rng.normal(80, 15, 100)
store_b = rng.normal(110, 20, 100)
store_c = rng.normal(95, 12, 100)
store_d = rng.normal(130, 25, 100)

# Add a few potential outliers
store_a = np.append(store_a, [160, 175])
store_d = np.append(store_d, [220, 240])

spending = [store_a, store_b, store_c, store_d]
stores = ["Store A", "Store B", "Store C", "Store D"]

# 2. Create 1x2 subplots
fig, ax = plt.subplots(1, 2, figsize=(11, 4.5))

# LEFT: Basic box plot
ax[0].boxplot(spending)
ax[0].set_title("Customer Spending")
ax[0].set_xlabel("Stores")
ax[0].set_ylabel("Spending")

# RIGHT: Improved LaTeX-style box plot
bp = ax[1].boxplot(
    spending,
    patch_artist=True,
    tick_labels=stores,
    widths=0.5,
    medianprops={"color": "black", "linewidth": 2},
    boxprops={"edgecolor": "black", "linewidth": 1.2},
    whiskerprops={"color": "black", "linewidth": 1.2},
    capprops={"color": "black", "linewidth": 1.2},
    flierprops={
        "marker": "o",
        "markerfacecolor": "red",
        "markeredgecolor": "black",
        "markersize": 5,
        "alpha": 0.7
    }
)

colors = ["royalblue", "mediumseagreen", "gold", "lightcoral"]

for box, color in zip(bp["boxes"], colors):
    box.set_facecolor(color)
    box.set_alpha(0.75)

ax[1].set_title(r"$\mathrm{Customer\ Spending\ by\ Store}$",
                fontsize=16, pad=12)

ax[1].set_xlabel(r"$\mathrm{Stores}$", fontsize=13)
ax[1].set_ylabel(r"$\mathrm{Customer\ Spending}$", fontsize=13)

ax[1].grid(axis="y", linestyle="--", alpha=0.3)
ax[1].set_axisbelow(True)

# Complete black box
for spine in ax[1].spines.values():
    spine.set_visible(True)
    spine.set_color("black")
    spine.set_linewidth(1.2)

# 3. Display
plt.tight_layout()
plt.show()
```
