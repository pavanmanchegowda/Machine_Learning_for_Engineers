# $\mathrm{1.0\ Introduction\ to\ Git\ and\ GitHub}$

## $\mathrm{1.1\ The\ Problem:\ Keeping\ Track\ of\ Changes}$

Without version control, a project folder quickly looks like this:

```text
analysis.ipynb
analysis_v2.ipynb
analysis_v2_corrected.ipynb
analysis_final.ipynb
analysis_final_REALLY_final.ipynb
analysis_final_Hubert_comments.ipynb
```

$\displaystyle
\begin{aligned}
&\text{Which file is the current one? What changed between two versions, and why?}\\[5pt]
&\text{How do two people work on the same code without overwriting each other?}\\[5pt]
&\text{How do you get back a version that worked yesterday?}
\end{aligned}
$

## $\mathrm{1.2\ What\ is\ Version\ Control?}$

$\displaystyle
\begin{aligned}
&\text{A version control system records every change to a set of files over time.}\\[5pt]
&\text{Every saved state (a commit) has an author, a date and a message explaining the change.}\\[5pt]
&\text{Any earlier state can be restored at any time.}\\[5pt]
&\text{Several people can work in parallel and combine their changes.}
\end{aligned}
$

> **Analogy:** a version control system works like the transaction log of a database: every change is recorded in order, can be traced to its author, and earlier states can be restored.

## $\mathrm{1.3\ Git\ and\ GitHub\ -\ Two\ Different\ Things}$

| | Git | GitHub |
|---|---|---|
| What is it? | A version control **program** | A **website** that hosts Git repositories |
| Where does it run? | On your computer (also in Colab) | In the cloud (github.com) |
| Created | 2005 by Linus Torvalds (for the Linux kernel) | 2008, owned by Microsoft since 2018 |
| Needs internet? | No – works fully offline | Yes |
| Main purpose | Record and manage versions of your files | Share code, collaborate, publish, back up |

$\displaystyle
\begin{aligned}
&\text{Git is the tool. GitHub is one of several platforms to store Git repositories online.}\\[5pt]
&\text{Alternatives to GitHub: GitLab, Bitbucket, or a self-hosted server.}
\end{aligned}
$

## $\mathrm{1.4\ Key\ Terms}$

| Term | Meaning |
|---|---|
| **Repository (repo)** | A project folder whose complete history is tracked by Git |
| **Commit** | A saved snapshot of the project, with author, date and message |
| **Working directory** | The files as you currently see and edit them |
| **Staging area** | A "waiting room" where you collect the changes for the next commit |
| **Branch** | An independent line of development (default branch: `main`) |
| **Remote** | A copy of the repository on a server, e.g. on GitHub (usually called `origin`) |
| **Clone** | Download a complete copy of a remote repository to your computer |
| **Push** | Upload your new commits to the remote repository |
| **Pull** | Download new commits from the remote repository and merge them into yours |
| **Pull request** | On GitHub: a request to merge changes from one branch into another, with review |

## $\mathrm{1.5\ The\ Three\ Areas\ of\ Git}$

```text
  Working directory  --git add-->  Staging area  --git commit-->  Local repository  --git push-->  GitHub (remote)
        ^                                                                                              |
        +--------------------------------------------  git pull  -------------------------------------+
```

$\displaystyle
\begin{aligned}
&\text{1. You edit files in the working directory.}\\[5pt]
&\text{2. git add selects which changes go into the next snapshot.}\\[5pt]
&\text{3. git commit saves the snapshot in your local repository - still only on your computer.}\\[5pt]
&\text{4. git push uploads your commits to GitHub; git pull downloads the commits of others.}
\end{aligned}
$

# $\mathrm{3.0\ Git\ in\ Practice:\ A\ Complete\ Example}$

$\displaystyle
\begin{aligned}
&\text{We now create a small repository inside this notebook and go through the basic workflow.}\\[5pt]
&\text{Every command is executed for real - you can run the cells yourself, also in Colab.}\\[5pt]
&\text{Each cell starts with cd git_demo, because every ! command runs in a new shell.}

## $\mathrm{3.1\ Creating\ a\ Repository:\ git\ init}$

````python
# Remove an old demo folder (if the notebook is run twice) and create a new, empty one
!rm -rf git_demo && mkdir git_demo

# git init turns the folder into a Git repository; -b main names the first branch "main"
!cd git_demo && git init -b main

# Set name and e-mail only for this demo repository (without --global)
!cd git_demo && git config user.name "Student Demo" && git config user.email "student@example.com"
````

Git has created a hidden folder `.git` inside `git_demo`. It contains the complete history of the project.
**Never edit or delete this folder by hand.**

````python
# -a also lists hidden files and folders (names starting with a dot)
!ls -a git_demo
````

## $\mathrm{3.2\ Checking\ the\ State:\ git\ status}$

`git status` is the most important command: it tells you which files are new, changed or ready to be committed.

````python
# Create a first file with Python code (writefile-like: echo writes the text into the file)
!cd git_demo && echo 'print("Hello, Git!")' > hello.py

# Ask Git about the state of the repository
!cd git_demo && git status
````

Git found the new file `hello.py`, but lists it as **untracked**: Git does not track it yet.

## $\mathrm{3.3\ Staging\ Changes:\ git\ add}$

````python
# Move hello.py into the staging area = "this file belongs to the next commit"
!cd git_demo && git add hello.py

# The file is now listed under "Changes to be committed"
!cd git_demo && git status
````

## $\mathrm{3.4\ Saving\ a\ Snapshot:\ git\ commit}$

A commit saves everything in the staging area as one snapshot. The message after `-m` explains **what** changed and **why**.

````python
# Create the first commit with a short, meaningful message
!cd git_demo && git commit -m "Add hello.py with first print statement"
````

## $\mathrm{3.5\ Changing\ a\ File\ and\ Viewing\ the\ Difference:\ git\ diff}$

````python
# Append a second line to hello.py (>> appends, > would overwrite)
!cd git_demo && echo 'print("Version control is useful.")' >> hello.py

# git diff shows the changes that are not yet staged:
#   lines starting with + were added, lines starting with - were removed
!cd git_demo && git diff
````

````python
# Stage and commit the change in one go
!cd git_demo && git add hello.py && git commit -m "Add second message to hello.py"

# Add a data file in a separate commit
!cd git_demo && printf "Temperature,Pressure\n25,1.0\n30,1.5\n" > experiment.csv
!cd git_demo && git add experiment.csv && git commit -m "Add experiment data"
````

## $\mathrm{3.6\ Viewing\ the\ History:\ git\ log}$

````python
# --oneline: one line per commit (short ID + message), newest first
!cd git_demo && git log --oneline
````

Every commit has a unique ID (the hexadecimal code on the left). It is used to refer to that exact version of the project.

## $\mathrm{3.7\ Undoing\ a\ Mistake:\ git\ restore}$

````python
# Oops - we accidentally overwrite hello.py with wrong content
!cd git_demo && echo "this is a mistake" > hello.py
!cd git_demo && cat hello.py

# git restore brings the file back to the state of the last commit
!cd git_demo && git restore hello.py
!echo "--- after git restore ---"
!cd git_demo && cat hello.py
````

## $\mathrm{3.8\ Ignoring\ Files:\ .gitignore}$

$\displaystyle
\begin{aligned}
&\text{Some files should never be committed: passwords, large datasets, temporary files.}\\[5pt]
&\text{A file named .gitignore lists files and folders that Git should ignore.}\\[5pt]
&\text{GitHub offers ready-made .gitignore templates, e.g. for Python.}
\end{aligned}
$

````python
# Write a simple .gitignore file: ignore notebook checkpoints, Python caches and secret files
!cd git_demo && printf ".ipynb_checkpoints/\n__pycache__/\n*.env\n" > .gitignore

# Create a file with a (fake) password - it must never end up on GitHub
!cd git_demo && echo "API_KEY=secret123" > passwords.env

# passwords.env does not appear in the status: it is ignored. Only .gitignore itself is new
!cd git_demo && git status --short
!cd git_demo && git add .gitignore && git commit -m "Add .gitignore"
````

## $\mathrm{3.9\ Branches}$

$\displaystyle
\begin{aligned}
&\text{A branch is a separate line of development.}\\[5pt]
&\text{You can try out a new idea on a branch without changing the working version on main.}\\[5pt]
&\text{If the idea works, the branch is merged back into main.}
\end{aligned}
$

````python
# Create a new branch and switch to it in one step
!cd git_demo && git switch -c new-plot

# Work on the branch: add a plotting script and commit it
!cd git_demo && echo 'import matplotlib.pyplot as plt' > plot.py
!cd git_demo && git add plot.py && git commit -m "Add plotting script"

# Go back to main - plot.py does not exist here yet
!cd git_demo && git switch main && ls

# Merge the branch into main - now plot.py is part of main
!cd git_demo && git merge new-plot && ls
````

````python
# The complete history as a graph: --all shows all branches
!cd git_demo && git log --oneline --graph --all
````

## $\mathrm{4.3\ Sending\ and\ Receiving\ Changes:\ git\ push\ and\ git\ pull}$

```bash
git pull                    # 1. start of work: get the latest changes from GitHub
# ... edit files ...
git status                  # 2. what has changed?
git add analysis.ipynb      # 3. stage the changes
git commit -m "Add correlation plot for concrete data"   # 4. save a snapshot locally
git push                    # 5. upload the new commits to GitHub
```

$\displaystyle
\begin{aligned}
&\text{Always pull before you start working - otherwise your push may be rejected.}\\[5pt]
&\text{If git push is rejected, somebody else pushed first: run git pull, then push again.}
\end{aligned}
$

## $\mathrm{4.4\ Logging\ In:\ Authentication}$

Since 2021, GitHub no longer accepts your account password for `git push` on the command line. Instead, use one of these:

| Method | When to use |
|---|---|
| **Browser login via Git Credential Manager** | Windows (included in Git for Windows): a browser window opens at the first push |
| **VS Code: Sign in with GitHub** | VS Code asks you to log in in the browser – the easiest way |

# $\mathrm{5.0\ Git\ in\ VS\ Code}$

# $\mathrm{7.0\ Good\ Practice\ and\ Common\ Problems}$

## $\mathrm{7.1\ Good\ Commit\ Messages}$

| Bad | Good |
|---|---|
| `update` | `Add train/test split to regression notebook` |
| `fix` | `Fix wrong column name in correlation plot` |
| `asdf` | `Remove unused imports` |

$\displaystyle
\begin{aligned}
&\text{Commit small, logical steps - one change per commit.}\\[5pt]
&\text{Write the message in the imperative: Add, Fix, Remove, Update.}
\end{aligned}
$

## $\mathrm{7.2\ What\ Never\ Belongs\ in\ a\ Repository}$

$\displaystyle
\begin{aligned}
&\text{Passwords, API keys and access tokens - deleting them later does not remove them from the history.}\\[5pt]
&\text{Personal or confidential company data.}\\[5pt]
&\text{Very large files: GitHub warns from 50 MB and blocks files larger than 100 MB.}\\[5pt]
&\text{Automatically generated files: .ipynb_checkpoints, __pycache__.}
\end{aligned}
$

## $\mathrm{7.3\ Merge\ Conflicts}$

If two people change the **same lines** of a file, Git cannot decide which version is correct. It marks the conflict in the file:

```text
<<<<<<< HEAD
plt.title("Strength vs. age")
=======
plt.title("Compressive strength over time")
>>>>>>> new-plot
```

$\displaystyle
\begin{aligned}
&\text{Edit the file: keep the correct version and delete the marker lines.}\\[5pt]
&\text{Then git add the file and git commit - the conflict is resolved.}\\[5pt]
&\text{VS Code shows buttons to accept one version or both.}
\end{aligned}
$

> **Notebooks and conflicts:** `.ipynb` files contain outputs and metadata, so conflicts in notebooks are hard to read. Avoid two people editing the same notebook at the same time.

# $\mathrm{8.0\ Summary:\ Git\ Cheat\ Sheet}$

| Command | What it does |
|---|---|
| `git config --global user.name "Name"` | set your name (once per computer) |
| `git init` | turn the current folder into a repository |
| `git clone <URL>` | download a repository from GitHub |
| `git status` | show new, changed and staged files |
| `git add <file>` / `git add .` | stage one file / all changes |
| `git commit -m "message"` | save a snapshot of the staged changes |
| `git log --oneline` | show the history |
| `git diff` | show unstaged changes line by line |
| `git restore <file>` | discard changes in a file (back to the last commit) |
| `git pull` | get and merge new commits from GitHub |
| `git push` | upload your commits to GitHub |
| `git switch -c <branch>` | create a new branch and switch to it |
| `git switch main` | go back to the main branch |
| `git merge <branch>` | merge a branch into the current branch |

## $\mathrm{8.1\ Checkpoint\ Questions}$

1. What is the difference between Git and GitHub?
   <details><summary>Answer</summary>Git is the version control program that runs on your computer. GitHub is a website that stores Git repositories online for sharing and collaboration.</details>

2. You changed `analysis.ipynb`. Which three commands bring the change to GitHub?
   <details><summary>Answer</summary><code>git add analysis.ipynb</code>, <code>git commit -m "…"</code>, <code>git push</code>.</details>

3. Your `git push` is rejected. What is the most likely reason, and what do you do?
   <details><summary>Answer</summary>Someone else pushed new commits first. Run <code>git pull</code> to merge them, then <code>git push</code> again.</details>

4. Why should a file with passwords never be committed, even if you delete it in the next commit?
   <details><summary>Answer</summary>Every commit stays in the history. Anyone with access to the repository can restore the old commit and read the password.</details>

5. What is the staging area good for?
   <details><summary>Answer</summary>It lets you choose exactly which changes go into the next commit, so one commit contains one logical change.</details>
