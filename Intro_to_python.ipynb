## $\mathrm{1.1\ What\ is\ Python?}$

$\displaystyle
\begin{aligned}
&\text{Python is a high-level, general-purpose, interpreted programming language.}\\[6pt]
&\text{High-level: Simple and readable syntax.}\\[4pt]
&\text{General-purpose: Used for different applications.}\\[4pt]
&\text{Interpreted: Code is executed by an interpreter.}\\[4pt]
&\text{Dynamically typed: Variable types are determined at runtime.}\\[4pt]
&\text{Open-source: Freely available and modifiable.}
\end{aligned}
$

## $\mathrm{1.3\ How\ Python\ Works}$

$\displaystyle
\begin{aligned}
&\text{The standard implementation of Python is called CPython.}\\[6pt]
&\text{Step 1: Write Python source code (.py).}\\[4pt]
&\text{Step 2: CPython compiles source code into bytecode.}\\[4pt]
&\text{Step 3: The Python Virtual Machine executes the bytecode.}\\[4pt]
&\text{Step 4: The program produces the output.}
\end{aligned}
$

$\displaystyle
\text{Source Code}
\longrightarrow
\text{Bytecode}
\longrightarrow
\text{Execution}
\longrightarrow
\text{Output}
$

## $\mathrm{1.8\ Python\ Script\ vs.\ Jupyter\ Notebook}$

$\displaystyle
\begin{aligned}
&\text{Python Script (.py):}\\[4pt]
&\quad\text{1. Contains Python code in a text file.}\\[3pt]
&\quad\text{2. Usually executed as a complete program.}\\[3pt]
&\quad\text{3. Suitable for automation, reusable functions, and applications.}\\[3pt]
&\quad\text{4. Output is displayed in the terminal or another configured destination.}\\[8pt]
&\text{Jupyter Notebook (.ipynb):}\\[4pt]
&\quad\text{1. Contains code cells, Markdown text, and visualizations.}\\[3pt]
&\quad\text{2. Individual cells can be executed independently.}\\[3pt]
&\quad\text{3. Variables remain in memory while the kernel is running.}\\[3pt]
&\quad\text{4. Suitable for learning, data analysis, and scientific experiments.}
\end{aligned}
$

### $\mathrm{Important\ Difference:\ Execution}$

$\displaystyle
\begin{aligned}
&\text{Script: Execution normally starts from the beginning of the file.}\\[4pt]
&\text{Notebook: Cells can be executed in any order.}\\[4pt]
&\text{Notebook results may depend on variables created in previously executed cells.}\\[4pt]
&\text{Restarting the kernel clears variables stored in memory.}
\end{aligned}
$

## $\mathrm{1.9\ How\ to\ Save\ Python\ Code}$

$\displaystyle
\begin{aligned}
&\text{Python scripts are saved using the .py extension.}\\[4pt]
&\text{Jupyter notebooks are saved using the .ipynb extension.}\\[6pt]
&\text{In VS Code, press Cmd + S on macOS or Ctrl + S on Windows.}\\[4pt]
&\text{Scripts save source code as plain text.}\\[4pt]
&\text{Notebooks save code, Markdown, and optionally execution outputs.}
\end{aligned}
$
