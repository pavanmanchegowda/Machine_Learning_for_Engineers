## $\mathrm{2.3\ Engineering\ Example}$


# $\mathrm{5.0\ Selecting\ Rows\ and\ Columns}$

$\displaystyle
\begin{aligned}
&\text{Pandas allows us to extract specific rows and columns.}\\[5pt]
&\text{A single column can be selected using its name.}\\[5pt]
&\text{Multiple columns can be selected using a list of names.}
\end{aligned}
$

## $\mathrm{5.1\ loc\ vs.\ iloc}$

$\displaystyle
\begin{aligned}
&\texttt{loc: Select data using row and column labels.}\\[5pt]
&\texttt{iloc: Select data using integer positions.}\\[5pt]
&\text{Integer positions begin at zero.}\\[5pt]
&\text{Label slices in loc include the final label.}\\[5pt]
&\text{Position slices in iloc exclude the final position.}
\end{aligned}
$


```python

# One column
print(df["Temperature"])

# Multiple columns
display(df[["Temperature", "Pressure"]])

# First row
print(df.iloc[0])

# First two rows
display(df.iloc[0:2])

# Specific value by label
print(df.loc[0, "Temperature"])
```


# $\mathrm{6.0\ Filtering\ Data}$

$\displaystyle
\begin{aligned}
&\text{Filtering selects rows that satisfy specified conditions.}\\[5pt]
&\text{For example, we may select measurements above a temperature limit.}\\[5pt]
&\text{Multiple conditions can be combined using logical operators.}
\end{aligned}
$

## $\mathrm{6.1\ Comparison\ Operators}$

$\displaystyle
\begin{aligned}
&> \quad \text{Greater than}\\[5pt]
&< \quad \text{Less than}\\[5pt]
&\geq \quad \text{Greater than or equal to}\\[5pt]
&\leq \quad \text{Less than or equal to}\\[5pt]
&= \quad \text{Equal to (Python: ==)}
\end{aligned}
$

$\displaystyle
\begin{aligned}
&\text{In Pandas, \& combines conditions using AND.}\\[5pt]
&\text{The vertical bar combines conditions using OR.}\\[5pt]
&\text{Each condition must be enclosed in parentheses.}
\end{aligned}
$


```python

# Temperature greater than 25
display(df[df["Temperature"] > 25])

# Pressure greater than or equal to 1.5
display(df[df["Pressure"] >= 1.5])

# Two conditions
filtered = df[
    (df["Temperature"] > 25) &
    (df["Pressure"] >= 1.5)
]

display(filtered)
```

## $\mathrm{7.2\ Engineering\ Example:\ Temperature\ Conversion}$

$\displaystyle
\begin{aligned}
&\text{Consider temperature measurements recorded in degrees Celsius.}\\[5pt]
&\text{We want to convert all measurements into Kelvin.}
\end{aligned}

$\displaystyle T_K=T_C+273.15$

$\displaystyle

$\displaystyle
\begin{aligned}
&\text{Experimental datasets may contain missing measurements.}\\[5pt]
&\text{Pandas commonly represents missing numerical values using NaN.}\\[5pt]
&\text{Missing data must be handled before certain numerical analyses.}
\end{aligned}
$

## $\mathrm{8.1\ Important\ Functions}$

$\displaystyle
\begin{aligned}
&\texttt{isna(): Identify missing values.}\\[5pt]
&\texttt{isna().sum(): Count missing values.}\\[5pt]
&\texttt{dropna(): Remove rows containing missing values.}\\[5pt]
&\texttt{fillna(): Replace missing values.}
\end{aligned}
$

## $\mathrm{8.2\ Engineering\ Example}$

$\displaystyle
\begin{aligned}
&\text{A temperature sensor may fail to record a measurement.}\\[5pt]
&\text{One option is to remove the incomplete measurement.}\\[5pt]
&\text{Another option is to estimate the missing value.}\\[5pt]
&\text{Mean replacement is simple but is not suitable for every dataset.}
\end{aligned}
$


```python

data_missing = {
    "Temperature": [25, np.nan, 35, 40],
    "Pressure": [1.0, 1.5, np.nan, 2.5]
}

df_missing = pd.DataFrame(data_missing)

display(df_missing)

print("Missing values:")
print(df_missing.isna().sum())

# Remove missing data
display(df_missing.dropna())

# Fill missing values with column means
df_filled = df_missing.fillna(df_missing.mean())

display(df_filled)
```


# $\mathrm{10.0\ Grouping\ Data}$

$\displaystyle
\begin{aligned}
&\text{Grouping divides a dataset into categories.}\\[5pt]
&\text{Statistical calculations can then be performed separately for each group.}\\[5pt]
&\text{The groupby() function is used for this purpose.}
\end{aligned}
$

## $\mathrm{10.1\ Engineering\ Example}$

$\displaystyle
\begin{aligned}
&\text{Consider temperature and pressure measurements from two machines.}\\[5pt]
&\text{We want to calculate the average measurements for each machine.}\\[5pt]
&\text{Pandas groups the rows by machine and calculates the mean.}
\end{aligned}
$


```python

data = {
    "Machine": ["A", "A", "B", "B", "A", "B"],
    "Temperature": [30, 35, 40, 45, 32, 42],
    "Pressure": [1.2, 1.4, 1.6, 1.8, 1.3, 1.7]
}

df_machines = pd.DataFrame(data)

display(df_machines)

result = df_machines.groupby("Machine")[
    ["Temperature", "Pressure"]
].mean()

display(result)
```

## $\mathrm{11.2\ Example:\ Stress\ Calculation}$

$\displaystyle
\begin{aligned}
&\text{Consider a bar subjected to an axial force.}\\[5pt]
&\text{Normal stress is calculated using the following equation.}
\end{aligned}
$

$\displaystyle \sigma=\frac{F}{A}$

$\displaystyle
\begin{aligned}
&F=\text{Applied force (N)}\\[5pt]
&A=\text{Cross-sectional area (m}^2\text{)}\\[5pt]
&\sigma=\text{Normal stress (Pa)}
\end{aligned}
$


```python

import numpy as np

df_example = pd.DataFrame({
    "Force_N": [100, 200, 300],
    "Area_m2": [0.002, 0.004, 0.005]
})

# Pandas to NumPy
force = df_example["Force_N"].to_numpy()
area = df_example["Area_m2"].to_numpy()

# Calculate stress
stress = force / area

# NumPy results back to Pandas
df_example["Stress_Pa"] = stress

display(df_example)
```
